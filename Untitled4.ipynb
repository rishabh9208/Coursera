```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import folium

%matplotlib inline

# Load dataset
URL = "https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBMDeveloperSkillsNetwork-DV0101EN-SkillsNetwork/Data%20Files/historical_automobile_sales.csv"
df = pd.read_csv(URL)
```

```python
# Group by Year and calculate mean automobile sales
df_line = df.groupby('Year')['Automobile_Sales'].mean()

plt.figure(figsize=(10, 6))
df_line.plot(kind='line')
plt.title('Automobile Sales Fluctuation from Year to Year')
plt.xlabel('Year')
plt.ylabel('Average Automobile Sales')
plt.grid(True)
plt.show()
```

```python
df_Mline = df.groupby(['Year', 'Vehicle_Type'], as_index=False)['Automobile_Sales'].sum()

plt.figure(figsize=(10, 6))
sns.lineplot(data=df_Mline, x='Year', y='Automobile_Sales', hue='Vehicle_Type')
plt.title('Automobile Sales Trend by Vehicle Type Over the Years')
plt.xlabel('Year')
plt.ylabel('Automobile Sales')
plt.legend(title='Vehicle Type', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.grid(True)
plt.show()
```

```python
df_rec = df.groupby('Recession')['Automobile_Sales'].mean().reset_index()
df_rec['Recession'] = df_rec['Recession'].map({0: 'Non-Recession', 1: 'Recession'})

plt.figure(figsize=(8, 6))
sns.barplot(data=df_rec, x='Recession', y='Automobile_Sales')
plt.title('Average Automobile Sales: Recession vs. Non-Recession Periods')
plt.xlabel('Period')
plt.ylabel('Average Automobile Sales')
plt.show()
```

```python
plt.figure(figsize=(10, 6))
sns.barplot(data=df, x='Vehicle_Type', y='Automobile_Sales', hue='Recession')
plt.title('Automobile Sales by Vehicle Type: Recession vs Non-Recession')
plt.xlabel('Vehicle Type')
plt.ylabel('Automobile Sales')
plt.legend(title='Recession', labels=['Non-Recession (0)', 'Recession (1)'])
plt.xticks(rotation=45)
plt.show()
```

```python
rec_data = df[df['Recession'] == 1]
non_rec_data = df[df['Recession'] == 0]

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 6))

sns.lineplot(data=rec_data, x='Year', y='GDP', ax=ax1, label='Recession GDP', color='red')
ax1.set_title('GDP Variation During Recession Periods')
ax1.set_xlabel('Year')
ax1.set_ylabel('GDP')

sns.lineplot(data=non_rec_data, x='Year', y='GDP', ax=ax2, label='Non-Recession GDP', color='blue')
ax2.set_title('GDP Variation During Non-Recession Periods')
ax2.set_xlabel('Year')
ax2.set_ylabel('GDP')

plt.tight_layout()
plt.show()
```

```python
rec_data = df[df['Recession'] == 1]

plt.figure(figsize=(10, 6))
sns.scatterplot(data=rec_data, x='Consumer_Confidence', y='Automobile_Sales')
plt.title('Consumer Confidence vs Automobile Sales During Recession Periods')
plt.xlabel('Consumer Confidence Index')
plt.ylabel('Automobile Sales')
plt.grid(True)
plt.show()
```

```python
rec_data = df[df['Recession'] == 1]
non_rec_data = df[df['Recession'] == 0]

rec_sales = rec_data.groupby('Vehicle_Type')['Automobile_Sales'].sum()
non_rec_sales = non_rec_data.groupby('Vehicle_Type')['Automobile_Sales'].sum()

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 6))

rec_sales.plot(kind='pie', ax=ax1, autopct='%1.1f%%', startangle=90)
ax1.set_title('Sales Distribution by Vehicle Type (Recession)')
ax1.set_ylabel('')

non_rec_sales.plot(kind='pie', ax=ax2, autopct='%1.1f%%', startangle=90)
ax2.set_title('Sales Distribution by Vehicle Type (Non-Recession)')
ax2.set_ylabel('')

plt.tight_layout()
plt.show()
```

```python
rec_data = df[df['Recession'] == 1]

plt.figure(figsize=(10, 6))
sns.countplot(data=rec_data, x='Vehicle_Type', hue='Engine_Size')
plt.title('Engine Size Distribution Across Vehicle Types During Recession')
plt.xlabel('Vehicle Type')
plt.ylabel('Count')
plt.legend(title='Engine Size', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.xticks(rotation=45)
plt.show()
```

```python
plt.figure(figsize=(10, 6))
sns.lineplot(data=df, x='Advertising_Expenditure', y='Automobile_Sales', hue='Recession')
plt.title('Advertising Expenditure vs Automobile Sales')
plt.xlabel('Advertising Expenditure')
plt.ylabel('Automobile Sales')
plt.legend(title='Period', labels=['Non-Recession (0)', 'Recession (1)'])
plt.grid(True)
plt.show()
```
