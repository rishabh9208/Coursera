```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import folium

%matplotlib inline

# Load the historical automobile sales dataset
URL = "https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBMDeveloperSkillsNetwork-DV0101EN-SkillsNetwork/Data%20Files/historical_automobile_sales.csv"
df = pd.read_csv(URL)

# TASK 1.1: Line chart showing automobile sales fluctuation year-to-year
df_line = df.groupby('Year')['Automobile_Sales'].mean()
plt.figure(figsize=(10, 6))
df_line.plot(kind='line')
plt.title('Automobile Sales Fluctuation from Year to Year')
plt.xlabel('Year')
plt.ylabel('Average Automobile Sales')
plt.grid(True)
plt.show()

# TASK 1.2: Line chart showing advertising expenditure correlate with sales in non-recession
df_non_rec = df[df['Recession'] == 0]
plt.figure(figsize=(10, 6))
sns.lineplot(data=df_non_rec, x='Advertising_Expenditure', y='Automobile_Sales')
plt.title('Advertising Expenditure vs Automobile Sales During Non-Recession Periods')
plt.xlabel('Advertising Expenditure')
plt.ylabel('Automobile Sales')
plt.grid(True)
plt.show()

# TASK 1.3: Bar chart comparing sales trend per vehicle type: recession vs non-recession
plt.figure(figsize=(10, 6))
sns.barplot(data=df, x='Vehicle_Type', y='Automobile_Sales', hue='Recession')
plt.title('Automobile Sales by Vehicle Type: Recession vs Non-Recession')
plt.xlabel('Vehicle Type')
plt.ylabel('Automobile Sales')
plt.legend(title='Recession', labels=['Non-Recession (0)', 'Recession (1)'])
plt.xticks(rotation=45)
plt.show()

# TASK 1.4: Subplots to compare variations in GDP during recession and non-recession periods
rec_data = df[df['Recession'] == 1]
non_rec_data = df[df['Recession'] == 0]
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 6))
sns.lineplot(data=rec_data, x='Year', y='GDP', ax=ax1, label='Recession GDP', color='red')
ax1.set_title('GDP Variation During Recession Periods')
ax1.set_xlabel('Year')
ax1.set_ylabel('GDP')
sns.lineplot(data=non_rec_data, x='Year', y='GDP', ax=ax2, label='Non-Recession GDP', color='blue')
ax2.set_title('GDP Variation During Non-Recession Periods')
ax2.set_xlabel('Year')
ax2.set_ylabel('GDP')
plt.tight_layout()
plt.show()

# TASK 1.5: Bubble plot for displaying the impact of seasonality on Automobile Sales
plt.figure(figsize=(10, 6))
sns.scatterplot(data=df, x='Month', y='Automobile_Sales', size='Seasonality_Weight', hue='Seasonality_Weight', sizes=(20, 400), legend=False)
plt.title('Impact of Seasonality on Automobile Sales')
plt.xlabel('Month')
plt.ylabel('Automobile Sales')
plt.grid(True)
plt.show()

# TASK 1.6: Scatter plot to identify correlation between average vehicle price and sales volume during recessions
rec_data = df[df['Recession'] == 1]
plt.figure(figsize=(10, 6))
sns.scatterplot(data=rec_data, x='Price', y='Automobile_Sales')
plt.title('Average Vehicle Price vs Sales Volume During Recessions')
plt.xlabel('Price')
plt.ylabel('Automobile Sales')
plt.grid(True)
plt.show()

# TASK 1.7: Pie chart displaying portion of advertising expenditure during recession and non-recession
rec_exp = df[df['Recession'] == 1]['Advertising_Expenditure'].sum()
non_rec_exp = df[df['Recession'] == 0]['Advertising_Expenditure'].sum()
plt.figure(figsize=(8, 6))
plt.pie([rec_exp, non_rec_exp], labels=['Recession', 'Non-Recession'], autopct='%1.1f%%', startangle=90)
plt.title('Advertising Expenditure: Recession vs Non-Recession')
plt.show()

# TASK 1.8: Pie chart displaying total Advertisement expenditure for each vehicle type during recession
rec_data = df[df['Recession'] == 1]
rec_ad = rec_data.groupby('Vehicle_Type')['Advertising_Expenditure'].sum()
plt.figure(figsize=(8, 6))
rec_ad.plot(kind='pie', autopct='%1.1f%%', startangle=90)
plt.title('Total Advertisement Expenditure Share by Vehicle Type During Recession')
plt.ylabel('')
plt.show()

# TASK 1.9: Line chart to analyze the effect of the unemployment rate on vehicle type and sales during Recession
rec_data = df[df['Recession'] == 1]
plt.figure(figsize=(10, 6))
sns.lineplot(data=rec_data, x='unemployment_rate', y='Automobile_Sales', hue='Vehicle_Type')
plt.title('Effect of Unemployment Rate on Vehicle Type and Sales During Recessions')
plt.xlabel('Unemployment Rate')
plt.ylabel('Automobile Sales')
plt.legend(title='Vehicle Type', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.grid(True)
plt.show()
```
